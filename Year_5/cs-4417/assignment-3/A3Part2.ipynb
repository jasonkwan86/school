```python
import torch
from transformers import AutoTokenizer, GPT2LMHeadModel

torch_device = "cuda" if torch.cuda.is_available() else "cpu"
tokenizer = AutoTokenizer.from_pretrained("gpt2")
# add the EOS token as PAD token to avoid warnings
model = GPT2LMHeadModel.from_pretrained("gpt2", pad_token_id=tokenizer.eos_token_id)
model_inputs = tokenizer('I am a student at Western University who loves to study math and machine learning. My favourite course is Unstructured Data 4417 and I like creating video games in my free time.', return_tensors='pt').to(torch_device)
torch.manual_seed(0)

print("My GPT-2 Story:")
print("---------------")
output = model.generate(
    **model_inputs,
    min_length=100,
    max_new_tokens=150,
    no_repeat_ngram_size=2
)

print(tokenizer.decode(
    output[0],
    skip_special_tokens=True
    ))
```
