```python
import json

def getTweets() -> list:
    tweets = []
    with open("tweets-utf-8.json", "r", encoding="utf-8") as file:
        for line in file:
            tweet = json.loads(line.strip())
            tweets.append(tweet["text"])
    return tweets
```

```python
import numpy as np
from operator import itemgetter

def sort_by_sim(query_embedding, document_embeddings, documents):
    data = []
    mag_query_embed = np.linalg.norm(query_embedding)
    for document_embedding, document in zip(document_embeddings, documents):
        mag_document_embed = np.linalg.norm(document_embedding)
        if mag_document_embed == 0:
            data.append((0, document))
        else:
            calculated_similarity = np.dot(query_embedding, document_embedding) / (mag_query_embed * mag_document_embed)
            data.append((calculated_similarity, document))
    data = sorted(data, key=itemgetter(0), reverse=True)
    return data
```

```python
from sentence_transformers import SentenceTransformer

def top25_model(query: str, documents, model: SentenceTransformer) -> list[str]:
    query_embedding = model.encode(query)
    document_embeddings = model.encode(documents)
    sorted_documents_by_simularity = sort_by_sim(query_embedding, document_embeddings, documents)

    return [x[1] for x in sorted_documents_by_simularity[:25]]
```

```python
from sentence_transformers import SentenceTransformer

glove_model = SentenceTransformer('sentence-transformers/average_word_embeddings_glove.840B.300d')

def top25_glove(query, documents):
    return top25_model(query, documents, glove_model)

print("**************GLOVE*****************")
for p in top25_glove("I am looking for a job.", getTweets()): print(p)
```

```python
from sentence_transformers import SentenceTransformer

minilm_model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

def top25_minilm(query, documents):
    return top25_model(query, documents, minilm_model)

print("**************MINILM*****************")
for p in top25_minilm("I am looking for a job.", getTweets()): print(p)
```
